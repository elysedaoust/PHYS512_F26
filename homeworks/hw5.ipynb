For this section, we know that $\mathbf{d}$ represents the data vector, and that it can be represented as $\mathbf{Aa}$, $\mathbf{A}$ is the matrix of which we will be getting the SVD, and $\mathbf{a}$ are the model parameters. For this problem, we are given the model $m(x) = a_1e^{-x} + a_2e^{-1.1x}$ to fit the observations. Given that the matrix $\mathbf{A}$ consists of the model basis functions as the row elements, and has as many rows as data points (which in this case is 10), we have:

$$
\mathbf{A} = \begin{bmatrix}
e^{-x_1} & e^{-1.1x_1} \\

e^{-x_10} & e^{-1.1x_10} \\

Let's use this matrix $\mathbf{A}$ to compute the SVD, and subsequently get our coefficients $a_1$ and $a_2$ for the model:

```python
import numpy as np
import matplotlib.pyplot as plt

# Define function
def f(x):
    return (np.exp(-x) + np.exp(-1.1*x))

# Function to get a1 and a2
def get_coeffs():
    # X values
    x = np.linspace(0,1,10)
    
    # Gaussian noise
    noise = np.random.normal(loc=0.0, scale=0.01, size=x.shape)
    
    # Data vector
    y = f(x) + noise
    
    # Matrix of basis functions
    A = np.column_stack((np.exp(-x), np.exp(-1.1*x)))
    
    # Compute SVD
    U, Sdiag, VT = np.linalg.svd(A, full_matrices=False)

    # Compute the sum from part a)
    u_t_1 = U.T[0,:]
    v_1 = VT.T[:,0]
    s_1 = Sdiag[0]
    c1 = (u_t_1 @ y)/s_1

    u_t_2 = U.T[1,:]
    v_2 = VT.T[:,1]
    s_2 = Sdiag[1]
    c2 = (u_t_2 @ y)/s_2

    # Get a1 and a2 from the sum
    a = c1*v_1 + c2*v_2
    a1 = a[0]
    a2 = a[1]

    return a1, a2

# Arrays to contain the model coefficients over multiple runs
a1 = []
a2 = []

# Loop to generate observations and fit to parameter
for i in range(0, 1000, 1):
    a_1, a_2 = get_coeffs()
    a1.append(a_1)
    a2.append(a_2)

plt.scatter(a1, a2, s=1)
plt.xlabel(r'$a_1$')
plt.ylabel(r'$a_2$')
plt.title(r'Measured values of $a_1$ and $a_2$')
plt.show()
```

Part a)

From the class notes, we know that the design matrix $\mathbf{A_{ij}}$ is given by:

$$
\mathbf{A_{ij}} = \frac{1}{\sigma_i} \frac{\partial m(t_i, \mathbf{a})}{\partial a_j}
$$

In this case, the model $m(x_i, \mathbf{a})$ is given by $rv(t, P, \mathbf{a})$, with $\mathbf{a} = (M_P, e, \omega_P, t_0, v_0)$.

As suggested, we can use finite differences to approximate the partial derivatives in the definition of $\mathbf{A_{ij}}$. With this, we can obtain an estimate for the partial derivatives with respect to each of the five parameters, by changing one of the parameters by a very small value and evaluating the model with the parameter before and after the change. For example, for the partial derivative with respect to the planet mass ($M_P$), we would get:

$$
\frac{\partial m}{\partial M_P} \approx \frac{m(t, P, M_P + h, e, \omega_P, t_0, v_0) - m(t, P, M_P, e, \omega_P, t_0, v_0)}{h},
$$

where in this case we would have $h = \Delta M_P$. The same process applies for the other four parameters (for each partial derivative, only one of the parameters changes). The design matrix is constructed in the function "design_matrix(t, P, a, errs)" below, following example code that was given in class.

```python
# Function to calculate the radial velocity (given in the problem)
def rv(t, P, a):
    # Calculates the radial velocity of a star orbited by a planet
    # at the times in the vector t
    
    # extract the orbit parameters
    # P, t and tp in days, mp in Jupiter masses, v0 in m/s  
    mp, e, omega, tp, v0 = a
        
    # mean anomaly
    M = 2*np.pi * (t-tp) / P
    
    # velocity amplitude
    K = 204 * P**(-1/3) * mp  / np.sqrt(1.0-e**2) # m/s
    
    # solve Kepler's equation for the eccentric anomaly E - e * np.sin(E) = M
    # Iterative method from Heintz DW, "Double stars", Reidel, 1978
    # first guess
    E = M + e*np.sin(M)  + ((e**2)*np.sin(2*M)/2)
    while True:
        E0 = E 
        M0 = E0 - e*np.sin(E0)
        E = E0 + (M-M0)/(1.0 - e*np.cos(E0))
        if np.max(np.abs((E-E0))) < 1e-6:
            break
        
    # evaluate the velocities
    theta = 2.0 * np.arctan( np.sqrt((1+e)/(1-e)) * np.tan(E/2))
    vel = v0 + K * ( np.cos(theta + omega) + e * np.cos(omega))
    
    return vel
```

```python
# Function to construct the design matrix (using parts of the example code in the Lorentzian fit exercise solutions)
def design_matrix(t, P, a, errs):
    # Define vector with slightly perturbed parameters (taken from class notes)
    eps = 1e-6
    b = a * (1+eps)

    A =  np.zeros([len(t), len(a)]) # initialize empty design matrix
    rv_ref = rv(t, P, a) # get value for the radial velocity without any perturbation in the parameters

    # Use rv function to calculate the model output with each slightly perturbed parameter (keeping all others fixed)
    rv1 = rv(t, P, [b[0], a[1], a[2], a[3], a[4]])
    rv2 = rv(t, P, [a[0], b[1], a[2], a[3], a[4]])
    rv3 = rv(t, P, [a[0], a[1], b[2], a[3], a[4]])
    rv4 = rv(t, P, [a[0], a[1], a[2], b[3], a[4]])
    rv5 = rv(t, P, [a[0], a[1], a[2], a[3], b[4]])

    # Construct design matrix, using finite differences (and dividing by the errors)
    A[:,0] = ((rv1 - rv_ref)/(b[0]-a[0]))/errs
    A[:,1] = ((rv2 - rv_ref)/(b[1]-a[1]))/errs
    A[:,2] = ((rv3 - rv_ref)/(b[2]-a[2]))/errs
    A[:,3] = ((rv4 - rv_ref)/(b[3]-a[3]))/errs
    A[:,4] = ((rv5 - rv_ref)/(b[4]-a[4]))/errs

    return rv_ref, A


# Levenberg-Marquardt (implemented using example in class notes)
def lev_mar(t, vobs, P, a0, errs):
    chisq1 = 1.0
    chisq = 1e99
    lam = 1e-3
    alpha = 0.2

    # keep going while chisq is dropping
    while chisq > 1e-3:
        # compute the update to the parameters
        f, A = design_matrix(t, P, a0, errs) # get model output and design matrix

        # Get the difference between the observations and the model
        r = (vobs - f)/errs

        chisq = np.sum(r**2)
        lhs = A.T@A
        lhs = lhs*(1 + np.identity(len(a0))*lam)
        rhs = A.T@r
        da = np.linalg.inv(lhs)@rhs
        # calculate the chisq associated with the new parameters
        a1 = a0 + alpha*da

        # Verify that the magnitude of the eccentricity is valid (between 0 and 1), if not reject immediately
        if (abs(a1[1]) >= 1):
            lam = lam*10
            continue

        f1 = rv(t, P, a1)
        r1 = (vobs - f1)/errs
        chisq1 = np.sum(r1**2)
        # print("chisq = %lg, a=(%lg, %lg, %lg), lam = %lg" % (chisq1 ,a1[0], a1[1], a1[2], lam))
        # accept if chisq decreases; reject if it increases
        if chisq1 > chisq:
            lam = lam * 10
        else:
            lam = lam / 10
            a0 = a1
            # if the improvement in chisq becomes too small then exit
            if chisq-chisq1 < 1e-3:
                break
    return a0
```

```python
P = 1724 # period in days

# Initial guess
a0 = np.array([1, 0.5, 1, 20, 1])

a_best_fit = lev_mar(tobs, vobs, P, a0, eobs)

print('Best fit parameters:')
print('M_P = ', a_best_fit[0], 'Jupiter masses')
print('e = ', a_best_fit[1])
print('ω_P = ', a_best_fit[2], 'rad')
print('t_0 = ', a_best_fit[3], 'days')
print('v_0 = ', a_best_fit[4], 'm/s')

chisq = np.sum(((vobs - rv(tobs, P, a_best_fit))/eobs)**2)
print("\nFinal chi squared = ", chisq)
```

## Question 3 - Photon random walk

Part a)

For this problem, we have an exponential distribution of the form $f(x) = \lambda e^{-\lambda x}$, with $\lambda = \frac{1}{l}$. This is very similar to the example explored in class for the transformation method. Choosing $y = -e^{-\lambda x}$, we have $\frac{dy}{dx} = \lambda e^{-\lambda x}$, so we can write:

$$
f(x)dx = \lambda e^{-\lambda x}dx = dy = f(y)dy,
$$

with $f(y) = 1$ (uniform distribution) and x ranging from 0 to $\infty$. For the transformation method, we choose a set of y values from the uniform distribution between the set limits (the corresponding limits for y are from -1 to 0), and then transform each one back to x using:

$$
x = -\frac{1}{\lambda}\ln(-y) = -l\ln(-y)
$$

Using this, let's choose $10^6$ random values for y, and transform them back to x:

```python
seed = 512
rng = np.random.default_rng(seed)

def transform(y, l):
    x = -l*np.log(-y)
    return x

def r_squared(x, y, z):
    return (x**2 + y**2 + z**2)

l = 1 # mean free path
c = 3e8 # speed of light, m/s
y = rng.uniform(-1, 0, 10**6) # 10^6 random generated values for y, uniformly distributed between -1 and 0

# Compute the step lengths between each scattering event
d = transform(y, l)
```

Now, we have our randomly generated step lengths given the established limits for y. Knowing that photon scattering is isotropic, let's now generate random directions for these steps. An important consideration is that there is equal probability for the photon to be scattered along a $\phi$ direction (azimuthal angle) between $0$ and $2\pi$, whereas for the $\theta$ direction (polar angle) there are a lot more possibilities near the equator of a sphere ($\theta = \frac{\pi}{2}$) compared to at the poles ($\theta = 0, \pi$), meaning that the probability is not equal for all values of $\theta$. Therefore, instead of using the individual angles, the probability of direction for the scattered photon therefore depends on the solid angle ($d\Omega$). In spherical coordinates, we have $d\Omega = \sin \theta d\theta d\phi$. Knowing that the probability of scattering into an infinitesimal solid angle $d\Omega$ is given by $dP = \frac{d\Omega}{4\pi}$, we have $dP = \frac{\sin \theta d\theta d\phi}{4\pi}$. From this, we can see that the probability distribution for $\theta$ is indeed not uniform. Using the substitution $\mu = \cos \theta$, giving $d\mu = -\sin \theta d\theta$, we can rewrite the probability as $dP = \frac{|d\mu| d\phi}{4\pi}$. From this, we can therefore see that we want uniform sampling of $\cos \theta$, and not of $\theta$. We will therefore sample $\phi$ between $0$ and $2\pi$, and $\mu = \cos \theta$ between $-1$ and $1$. This is used in the following code to get random uniformly distributed directions for the isotropic scattering of photons:

```python
# Generate uniformly distributed values for phi and mu (cos(theta))
phi = rng.uniform(0, 2*np.pi, 10**6)
mu = rng.uniform(-1, 1, 10**6)

# Compute distances travelled in each direction per step
theta = np.arccos(mu) # get values for theta
dx = d*np.sin(theta)*np.cos(phi)
dy = d*np.sin(theta)*np.sin(phi)
dz = d*np.cos(theta)

# Calculate the total displacement after each step
x_total = np.cumsum(dx)
y_total = np.cumsum(dy)
z_total = np.cumsum(dz)

# Calculate distance from the origin after each step
r2 = r_squared(x_total, y_total, z_total)

# Calculate time after each step
t = np.arange(l/c, (1e6 + 1)*(l/c), l/c)

# Plotting
plt.plot(t, r2)
plt.xlabel("Time (s)")
plt.ylabel(r"Squared distance from origin, $r^2$ ($m^2$)")
plt.title("Squared distance from the origin of a photon being scattered over time")
plt.show()
```

As expected, looking at the above plot, we see the behaviour of a normal walk, which describes the random isotropic scattering of a photon (many fluctuations in the squared displacement, with a general increasing trend).

Part b)

For this part, we will use the same process as part a), but repeated 100 times. Then, given that each row in the final array contains $r^2$ after each scattering for a particular trajectory, each column corresponds to the values of $r^2$ across all trajectories at a particular time. We can therefore compute the average of each column in order to obtain $<r^2>$ across all trajectories at any given time.

```python
def scatter(l, N):
    y = rng.uniform(-1, 0, N) # N random generated values for y, uniformly distributed between -1 and 0

    # Compute the step lengths between each scattering event
    d = transform(y, l)

    # Generate uniformly distributed values for phi and mu (cos(theta))
    phi = rng.uniform(0, 2*np.pi, N)
    mu = rng.uniform(-1, 1, N)

    # Compute distances travelled in each direction per step
    theta = np.arccos(mu) # get values for theta
    dx = d*np.sin(theta)*np.cos(phi)
    dy = d*np.sin(theta)*np.sin(phi)
    dz = d*np.cos(theta)

    # Calculate the total displacement after each step
    x_total = np.cumsum(dx)
    y_total = np.cumsum(dy)
    z_total = np.cumsum(dz)

    # Calculate distance from the origin after each step
    r2 = r_squared(x_total, y_total, z_total)
    return r2, x_total[-1]


# Define parameters
l = 1 # mean free path
c = 3e8 # speed of light, m/s
N = 10**6

# Initialize empty array that will contain all squared displacement values
r2_arr = np.zeros((100, N))

# Repeat part a) 100 times
for i in range(0,100,1):
    r_sq, _ = scatter(l, N)
    r2_arr[i] = r_sq
```

```python
# Compute column-wise average for values of r squared across all trajectories at any given time (since each column corresponds to a particular time)
avg = np.mean(r2_arr, axis=0)
t = np.arange(l/c, (N + 1)*(l/c), l/c)

# Predicted average from diffusion equation
avg_predicted = 6*(c*l/3)*t

# Compute relative error
err = np.abs((avg[-1] - avg_predicted[-1])/avg_predicted[-1])
print("Relative difference at final time:", err*100, '%')

# Plotting
plt.plot(t, avg, label='Numerically calculated average')
plt.plot(t, avg_predicted, label='Average calculated from prediction')
plt.xlabel("Time (s)")
plt.ylabel(r"Average squared distance from origin, $<r^2>$ ($m^2$)")
plt.title(r"Average value of $r^2$ across all trajectories at any given time")
plt.legend(loc='lower right')
plt.show()
```

Looking at the plot above, we can see that the numerical average follows the theoretical prediction quite closely, as expected. There are still small fluctuations for the average squared displacement of the photons, however we notice that with 100 averaged trials there are a lot less fluctuations than for the squared displacement for a single trial in part a). Comparing the numerical average to the theoretical prediction at the final time value, we find that the relative difference between the two is of around 4.27%. Therefore, this shows a good agreement between both calculated averages for $<r^2>$.

Part c)

Let's repeat the process in part b), but this time using $10^4$ trajectories, with 1000 steps each. This gives:

```python
# Define parameters
l = 1 # mean free path
c = 3e8 # speed of light, m/s
N = 1000

# Initialize empty array that will contain all squared displacement values
r2_arr_2 = np.zeros((10**4, N))
x_vals = np.zeros((10**4))

# Repeat part a) 100 times
for i in range(0,10**4,1):
    r_sq, x = scatter(l, N)
    r2_arr_2[i] = r_sq
    x_vals[i] = x
```

Now, let's plot a histogram of the final x-coordinate from each trajectory, and compare it with the Green's function for the diffusion equation:

$$
p(x) = \frac{1}{\sqrt{4\pi Dt}}e^{\frac{-x^2}{4Dt}}
$$

```python
# Define global variables, so that Green's function only takes 1 argument
t_final = t[-1] # final time value
D = c*l/3 # Diffusion coefficient

# Green's function for the diffusion equation
def green(x):
    p = (1/np.sqrt(4*np.pi*D*t_final))*np.exp(-(x**2)/(4*D*t_final))
    return p

# Plotting the histogram (function taken from class notes)
def plot_distribution(x, func, y_log=True):
    plt.clf()
    plt.hist(x, density=True, bins=100, histtype = 'step', label='Final x coordinates from trajectories')
    xx = np.linspace(min(x),max(x),100)
    plt.plot(xx, func(xx),':', label="Green's function")
    if y_log:
        plt.yscale('log')
    plt.xlabel('Final x coordinate (m)')
    plt.ylabel(r'Probability density ($m^{-1}$)')
    plt.title("Distribution of final x coordinates for photon trajectories")
    plt.legend(fontsize=7)
    plt.show()

plot_distribution(x_vals, green, y_log=False)
```

Therefore, from the plot above, we can tell that the distribution has the expected shape as predicted by Green's function for the diffusion equation (our histogram agrees well with the curve from Green's function, as can be observed given that it follows the curve closely). The distribution of final x coordinates looks like a Gaussian curve centered around x=0. This makes sense, since for photon scattering the photon has equal probability of being scattered in any direction (assuming isotropic scattering). However, we can still observe some random fluctuations, which is expected (especially since we are considering a finite number of trajectories), but the overall shape agrees well with the theoretical prediction nonetheless.
